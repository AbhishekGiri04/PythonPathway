```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

sns.lineplot(x=x, y=y)

plt.show()
```

```python
df=sns.load_dataset("tips")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.columns)
```

```python
sns.set_theme()
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip"
)
plt.title("Scatter Plot")
plt.show()
```

```python
sns.set_theme()
plt.figure(figsize=(8, 5))
sns.lineplot(
    data=df,
    x="total_bill",
    y="tip"
)
plt.title("Line Plot")
plt.show()
```

```python
sns.lineplot(
    data=df,
    x='day',
    y='total_bill',
    hue='smoker'
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    size="size"
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    style="time"
)

plt.show()
```

```python
sns.barplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.barplot(
    data=df,
    x="day",
    y="total_bill",
    estimator="sum"
)

plt.show()
```

```python
sns.countplot(
    data=df,
    x="day"
)

plt.show()
```

```python
sns.countplot(
    data=df,
    x="day",
    hue="sex"
)

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill"
)

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill",
    bins=10
)

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill",
    kde=True
)

plt.show()
```

```python
sns.kdeplot(
    data=df,
    x="total_bill"
)

plt.show()
```

```python
sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.violinplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.stripplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.swarmplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
corr = df.select_dtypes("number").corr()

sns.heatmap(corr)

plt.show()
```

```python
sns.heatmap(
    corr,
    annot=True
)

plt.show()
```

```python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.show()
```

```python
corr = df.select_dtypes("number").corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.show()
```

```python
sns.pairplot(df)

plt.show()
```

```python
sns.pairplot(
    df,
    hue="sex"
)

plt.show()
```

```python
sns.jointplot(
    data=df,
    x="total_bill",
    y="tip"
)

plt.show()
```

```python
sns.jointplot(
    data=df,
    x="total_bill",
    y="tip",
    kind="kde"
)

plt.show()
```
