```python
import matplotlib.pyplot as plt
```

```python
# 1. Create First Chart

# Data
x=[1,2,3,4,5]
y=[10,25,35,40,50]
plt.plot(x,y)
plt.grid()
plt.show()
```

```python
# 2. Customize Chart
plt.figure(figsize=(6,4))
plt.plot(x,y,color='b',marker='o',linestyle='dashed',linewidth=2,markersize=10)
plt.title("Random Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
# 3. Multiple Lines and Legends
# Data
x1=[1,2,3,4,5]
y1=[10,25,35,40,50]
y2=[20,35,45,50,60]
plt.plot(x1,y1,label="Sales 2025")
plt.plot(x1,y2,label="Sales 2026")
plt.title("Sales Chart")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# 4. Bar Chart
x=[1,2,3,4,5]
y=[10,25,35,40,50]
plt.bar(x,y,color='orange')
plt.title("Bar Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
# Horizontal-Bar Chart
x=[1,2,3,4,5]
y=[10,25,35,40,50]
plt.barh(x,y,color='pink')
plt.title("Horizontal-Bar Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
# 5. Histogram
# Used for distribution purpose
marks = [45, 52, 55, 58, 60, 61, 62, 65, 67, 68,
         70, 71, 72, 73, 74, 75, 76, 78, 80, 82,
         85, 87, 90, 92, 95]

plt.hist(marks,bins=5,edgecolor="black",color="purple")

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Violin Plot
data = [45, 52, 55, 58, 60, 61, 62, 65, 67, 68,
         70, 71, 72, 73, 74, 75, 76, 78, 80, 82,
         85, 87, 90, 92, 95]

plt.violinplot(data)

plt.title("Violin Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
# 6. Pie Chart
x = [10, 25, 35, 40, 50]
y = ["A", "B", "C", "D", "E"]

plt.pie(x,labels=y,autopct="%1.1f%%",startangle=90)
plt.title("Category Distribution")
plt.show()
```

```python
# 7. Scatter Plot
x=[1,2,3,4,5]
y=[10,25,35,40,50]

plt.scatter(x,y)
plt.title("Scatter Plot Chart")
plt.show()
```

```python
# 8. Subplots

# Data1
x1 = [10, 25, 35, 40, 50]
y1 = ["A", "B", "C", "D", "E"]

# Data2
x2 = [1,2,3,4,5]
y2 = [10,25,35,40,50]

plt.figure(figsize=(10,3))

plt.subplot(1,2,1) # row,column,position
plt.bar(x1,y1)
plt.title("Bar Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.subplot(1,2,2) # row,column,position
plt.scatter(x2,y2)
plt.title("Scatter Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.show()
```

```python
# 9. Matplotlib with pandas - real data
import pandas as pd

# data
data={
    'Months':['Jan','Feb','Mar','Apr'],
    'Sales':[1,2,3,4]
}

df=pd.DataFrame(data)
df
```

```python
plt.bar(df['Months'],df['Sales'])
plt.title("Matplotlib with Pandas")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.savefig("MWP.png") # 10. Save Charts
plt.show()
```
